QUESTION 3

 What probability distribution best fits significant wave height (Hs) data? Explain the physical justification and name the Python function to fit it.

ANSWER

The 3-parameter Weibull distribution (or 2-parameter Weibull for long-term ocean wave climate) is widely accepted as the standard probability distribution for fitting significant wave height ($H_s$) data. For extreme wave analysis, the Generalized Extreme Value (GEV) distribution or Generalized Pareto Distribution (GPD) are also frequently used.

PhysicalJustification

   1 Positive Asymmetry and Bounds: Significant wave height is strictly non-negative ($H_s \ge 0$) and inherently right-skewed. The Weibull distribution naturally enforces a lower bound at zero while modeling the long right tail of occasional high-energy sea states.
   2. Rayleigh / Wave Amplitude Basis: Short-term wave heights in a stationary sea state follow a Rayleigh distribution, which is a special case of the Weibull distribution (where the shape parameter $k = 2$). Aggregating short-term sea states over longer temporal scales (months/years) shifts the shape parameter ($k \approx 1.0 - 2.0$), making the flexible Weibull family ideal for capturing long-term variations.
    

 3.Threshold and Saturation Dynamics:

 Physical factors like wave breaking limits, fetch restrictions, and atmospheric energy transfers constrain maximum wave growth, which the shape ($k$), scale ($\lambda$), and location ($\gamma$) parameters of the 3-parameter Weibull model effectively capture.

```python
import numpy as np
from scipy.stats import weibull_min

# Sample significant wave height data (Hs)
hs_data = np.random.weibull(a=1.5, size=1000) * 2.0

# Fit the 3-parameter Weibull distribution: (shape, location, scale)
shape, loc, scale = weibull_min.fit(hs_data, floc=0)  # Fix loc=0 for 2-parameter fit

print(f"Shape (k): {shape:.4f}")
print(f"Location (gamma): {loc:.4f}")
print(f"Scale (lambda): {scale:.4f}")
```

 QUESTION 4

Explain the difference between the Block Maxima (GEV) and Peaks Over Threshold (GPD) methods for extreme value analysis. When would you prefer each?


 ANSWER

Both **Block Maxima (GEV)** and **Peaks Over Threshold (GPD)** are fundamental asymptotic methods in Extreme Value Theory (EVT), but they define and sample "extremes" differently.



1. Theoretical & Methodological Differences

| Feature | Block Maxima (GEV) | Peaks Over Threshold (GPD)

Underlying Theorem Fisher-Tippett-Gnedenko

 Theorem : Asserts that the normalized maxima of a sequence of independent and identically distributed (i.i.d.) random variables converge to the Generalized Extreme Value distribution. | **Pickands-Balkema-de Haan Theorem**: Asserts that the distribution of excesses over a sufficiently high threshold $u$ converges to the Generalized Pareto Distribution as the threshold approaches the right endpoint |


Data Selection Method** | Divide the time series into equal, non-overlapping time blocks (e.g., years, seasons) and extract **only the single maximum value** from each block. | Define a high physical threshold $u$, and select **all data points** that exceed this threshold ($X > u$). |

Sample Size / Data Use** | Low/Fixed. You get exactly $N$ data points for $N$ blocks, discarding other high values within any single block. | Flexible. Can extract many more data points because a single stormy period or block can contribute multiple extreme events. |
| **Parameters** | 3 parameters: Location ($\mu$), Scale ($\sigma$), and Shape ($\xi$). | 3 parameters: Threshold ($u$, chosen a priori), Scale ($\sigma_u$), and Shape ($\xi$). |



 2. When to Prefer Block Maxima (GEV)

You should prefer the **Block Maxima (GEV)** approach when:

* **The data is naturally partitioned and seasonal:** It is highly suitable for annual maxima (e.g., finding the 100-year return level for river discharges or wind speeds) where yearly cycles dictate natural boundaries.
* **Data reporting is low-resolution or pre-aggregated:** If you only have historical access to compiled annual maxima (which is common in legacy meteorological datasets) rather than continuous high-frequency raw time series.
* **Ease of avoiding temporal dependency:** Since block sizes (like 1 year) are typically large, the extracted block maxima are safely assumed to be independent and identically distributed (i.i.d.), avoiding complex de-clustering procedures.



 3. When to Prefer Peaks Over Threshold (GPD)

You should prefer the **Peaks Over Threshold (GPD)** approach when:

* **You have limited data / short record lengths:** GPD utilizes data much more efficiently. It prevents discarding secondary extreme events that occurred within the same block (e.g., if the second and third largest waves in a highly active year were still larger than the absolute maximum of a calm year).
* **The physical process is event-driven:** Marine storms, flood events, or market crashes happen in clusters. GPD models these physical "events" more dynamically.
* **You can robustly filter for independence:** Since consecutive observations above a threshold can be highly correlated (e.g., subsequent hours during the same storm), you must apply a de-clustering technique (like the run-length method) to ensure independent peak events.

The Jubilee TRP3 site has a 100-year Hs of approximately 3.8 m. Using the Ofon II operations threshold of 1.70 m/s current and the current profile from Table 3.7, what percentage of the water column exceeds this threshold during a 100-year event?


```python
import pandas as pd
import numpy as np

# Load the current profile dataset
file_path = '/content/jubilee_trp3_current_profile.csv'
df = pd.read_csv(file_path)

# Let's perform linear interpolation to find the exact depth where current_speed_ms is exactly 1.70
# We see it happens between 20m (1.75 m/s) and 30m (1.68 m/s)
depths = df['depth_m'].values
speeds = df['current_speed_ms'].values

target_speed = 1.70

# Linearly interpolate between the two points enclosing 1.70 m/s
x0, x1 = 20, 30
y0, y1 = 1.75, 1.68

# x = x0 + (y - y0) * (x1 - x0) / (y1 - y0)
exact_depth = x0 + (target_speed - y0) * (x1 - x0) / (y1 - y0)

max_depth = df['depth_m'].max() # Total depth of the water column
percentage_exceeded = (exact_depth / max_depth) * 100

print(f"The current drops below {target_speed} m/s at exactly {exact_depth:.2f} meters depth.")
print(f"The total water column depth modeled is {max_depth} meters.")
print(f"Percentage of the water column exceeding the threshold: {percentage_exceeded:.2f}%")
```
